```python
import pandas as pd
import numpy as np
```

```python
features = pd.read_csv("../data/elliptic_txs_features.csv", header=None)
features.head()
```

```python
features = features.rename(columns={0: "txId", 1: "time_step"})
```

```python
features.head()
```

```python
classes = pd.read_csv("../data/elliptic_txs_classes.csv")
classes.head()
```

```python
df = features.merge(classes, on="txId")
df.shape
```

```python
df.head()
```

```python
df["class"].value_counts()
```

```python
labeled = df[df["class"] != "unknown"]
```

```python
labeled.tail(20)
```

```python
labeled["target"] = np.where(labeled["class"] == "1", 1, 0)
```

```python
train = labeled[labeled["time_step"] <= 34]
```

```python
test = labeled[labeled["time_step"] > 34]
X_test, y_test = test.iloc[:, 2:167], test["target"]
```

```python
print(X_test.shape)
print(y_test.shape)
```

```python
train["time_step"].max()
test["time_step"].max()
```

```python
training = train[train["time_step"] <= 24]
validation = train[train["time_step"] > 24]
```

```python
training["time_step"].max()
validation["time_step"].max()
```

```python
X_train = training.iloc[:, 2:167]
y_train = training["target"]
X_train_val = validation.iloc[:, 2:167]
y_train_val = validation["target"]
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, classification_report

model_baseline = DummyClassifier()
model_baseline.fit(X_train, y_train)

y_predict = model_baseline.predict(X_train_val)
print(classification_report(y_train_val, y_predict))
```
